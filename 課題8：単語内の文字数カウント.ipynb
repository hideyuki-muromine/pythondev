```python
import string

words = []
counts = {ch: 0 for ch in string.ascii_lowercase}

while True:
    w = input("英単語を入力してください：")
    if w == "":
        break

    words.append(w)

    for ch in w.lower():
        if ch in counts:
            counts[ch] += 1

print("入力した英単語：", sorted(words))
print("合計カウント:", sum(counts.values()))  # ←確認用

for ch in string.ascii_lowercase:
    if counts[ch] > 0:
        print(f"{ch}が{counts[ch]}個ありました")
```
